# Import Data

```python
import os 
import pandas as pd
import numpy as np
```

```python
os.listdir("raw")
```

```python
statistics_dataframe = pd.DataFrame()
for file in os.listdir("raw/raw_statistics_parquet"):
    single_stats = pd.read_parquet("raw/raw_statistics_parquet/" + file)
    statistics_dataframe = pd.concat([statistics_dataframe, single_stats], axis = "rows", ignore_index = True)
```

```python
votes_dataframe = pd.DataFrame()
for file in os.listdir("raw/raw_votes_parquet"):
    single_votes = pd.read_parquet("raw/raw_votes_parquet/" + file)
    votes_dataframe = pd.concat([votes_dataframe, single_votes], axis = "rows", ignore_index= True)
```

```python
power_dataframe = pd.DataFrame()
for file in os.listdir("raw/raw_tennis_power_parquet"):
    single_power = pd.read_parquet("raw/raw_tennis_power_parquet/" + file)
    power_dataframe = pd.concat([power_dataframe, single_power], axis = "rows", ignore_index = True)
```

```python
odds_dataframe = pd.DataFrame()
for file in os.listdir("raw/raw_odds_parquet"):
    single_odds = pd.read_parquet("raw/raw_odds_parquet/" + file)
    odds_dataframe = pd.concat([odds_dataframe, single_odds], axis = "rows", ignore_index = True)
```

```python
pbp_dataframe = pd.DataFrame()
for file in os.listdir("raw/raw_point_by_point_parquet"):
    single_pbp = pd.read_parquet("raw/raw_point_by_point_parquet/" + file)
    pbp_dataframe = pd.concat([pbp_dataframe, single_pbp], axis = 0, ignore_index =True)
```

```python
away_team_dataframe = pd.DataFrame()
away_team_score_dataframe = pd.DataFrame()
event_dataframe = pd.DataFrame()
home_team_dataframe = pd.DataFrame()
home_team_score_dataframe = pd.DataFrame()
round_dataframe = pd.DataFrame()
season_dataframe = pd.DataFrame()
time_dataframe = pd.DataFrame()
tournament_dataframe = pd.DataFrame()
venue_dataframe = pd.DataFrame()


for file in os.listdir("raw/raw_match_parquet"):
    if file.startswith("away_team_1"):
        single_away = pd.read_parquet("raw/raw_match_parquet/" + file)
        away_team_dataframe = pd.concat([away_team_dataframe, single_away], axis = 0, ignore_index = True)
    if file.startswith("away_team_score"):
        single_away_team_score = pd.read_parquet("raw/raw_match_parquet/" + file)
        away_team_score_dataframe = pd.concat([away_team_score_dataframe, single_away_team_score], axis = 0, ignore_index = True)
    if file.startswith("event"):
        single_event = pd.read_parquet("raw/raw_match_parquet/" + file)
        event_dataframe = pd.concat([event_dataframe, single_event], axis = 0, ignore_index = True)
    if file.startswith("home_team_1"):
        single_home = pd.read_parquet("raw/raw_match_parquet/" + file)
        home_team_dataframe = pd.concat([home_team_dataframe, single_home], axis = 0, ignore_index = True)
    if file.startswith("home_team_score"):
        single_home_team_score = pd.read_parquet("raw/raw_match_parquet/" + file)
        home_team_score_dataframe = pd.concat([home_team_score_dataframe, single_home_team_score], axis = 0, ignore_index = True)
    if file.startswith("round"):
        single_round = pd.read_parquet("raw/raw_match_parquet/"+  file)
        round_dataframe = pd.concat([round_dataframe, single_round], axis = 0, ignore_index = True)
    if file.startswith("season"):
        single_season = pd.read_parquet("raw/raw_match_parquet/" + file)
        season_dataframe = pd.concat([season_dataframe, single_season], axis = 0, ignore_index = True)
    if file.startswith("time"):
        single_time = pd.read_parquet("raw/raw_match_parquet/" + file)
        time_dataframe = pd.concat([time_dataframe, single_time], axis = 0, ignore_index = True)
    if file.startswith("tournament"):
        single_tournament = pd.read_parquet("raw/raw_match_parquet/" + file)
        tournament_dataframe = pd.concat([tournament_dataframe, single_tournament], axis = 0, ignore_index = True)
    if file.startswith("venue"):
        single_venue = pd.read_parquet("raw/raw_match_parquet/" + file)
        venue_dataframe = pd.concat([venue_dataframe, single_venue], axis = 0, ignore_index = True)        
```

## Q8

> ### 8. Is there a difference in the number of double faults based on gender? 


### code

```python
double_faults_statistics = statistics_dataframe[statistics_dataframe["statistic_name"] == "double_faults"]
double_faults_statistics = double_faults_statistics[double_faults_statistics["period"] == "ALL"]
double_faults_statistics = double_faults_statistics.drop(["period", "statistic_category_name", "statistic_name", "compare_code"
                                , "statistic_type", "value_type", "home_total", "away_total", "home_stat", "away_stat"], axis = 1)
double_faults_statistics["double_faults_match"] = double_faults_statistics[["home_value", "away_value"]].sum(axis = 1)
double_faults_statistics = double_faults_statistics.drop(["home_value", "away_value"], axis = 1)

season_dataframe_men = season_dataframe.loc[season_dataframe["name"].str.contains("Men"), :]
season_dataframe_men["gender"] = "men"
season_dataframe_men = season_dataframe_men.reset_index().drop(["season_id", "name", "year", "index"], axis = 1)
season_dataframe_women = season_dataframe.loc[season_dataframe["name"].str.contains("Women"), :]
season_dataframe_women["gender"] = "women"
season_dataframe_women = season_dataframe_women.reset_index().drop(["season_id", "name", "year", "index"], axis = 1)
match_gender = pd.concat([season_dataframe_men, season_dataframe_women], axis = 0).reset_index().drop("index", axis = 1)
double_faults_gender = pd.merge(double_faults_statistics, match_gender, how = 'inner')
double_faults_gender.groupby("gender")["double_faults_match"].mean()
```


###
Answer
:

The average number of double faults per match for men's match is **5.56** and for women's match is **7.68**.It seems that women's do double faults **1.38** times more than men in a single match.


## Q9

> ### 9. Which player has won the most tournaments in a single month?

### codes

```python
event_dataframe["date"] = pd.to_datetime(event_dataframe["start_datetime"], unit = 's')
event_date = event_dataframe.drop(
    ["first_to_serve", "home_team_seed", "away_team_seed", "custom_id", "default_period_count", "final_result_only"], axis =1)

event_date["month"] = event_date["date"].dt.month


away_winner_ids = event_dataframe[event_dataframe['winner_code'] == 2].merge(away_team_dataframe, on='match_id')
home_winner_ids = event_dataframe[event_dataframe['winner_code'] == 1].merge(home_team_dataframe, on='match_id')
all_winner_ = pd.concat([home_winner_ids, away_winner_ids]).reset_index()
winner_per_match = all_winner_[["match_id", "slug", "name"]]

winner_month = pd.merge(winner_per_match, event_date)
winner_month = winner_month.drop(["start_datetime", "match_slug", "winner_code", "slug", "date"], axis  = 1)
winner_month_count = pd.DataFrame(winner_month.groupby(["name", "month"])["match_id"].count()).rename({"match_id" : "win_count"}, axis = 1)
winner_month_count.loc[winner_month_count["win_count"] == winner_month_count["win_count"].max(), :]
```


###
Answer
:

There are three players which has won the most matches in a single month. Each of these players has won 3 matches in a single month.


## Q12

> ### 12. What is the average number of games per set in men's matches compared to women's matches?


### codes

```python
average_games_perset_ =  power_dataframe.groupby(["match_id", "set_num"],)["game_num"].count().unstack().mean(axis = 1)
average_games_perset_ = pd.DataFrame(average_games_perset_).rename( {0 : "avg_games_per_set"}, axis = 1).reset_index()
season_dataframe_men = season_dataframe.loc[season_dataframe["name"].str.contains("Men"), :]
season_dataframe_men["gender"] = "men"
season_dataframe_men = season_dataframe_men.reset_index().drop(["season_id", "name", "year", "index"], axis = 1)
season_dataframe_women = season_dataframe.loc[season_dataframe["name"].str.contains("Women"), :]
season_dataframe_women["gender"] = "women"
season_dataframe_women = season_dataframe_women.reset_index().drop(["season_id", "name", "year", "index"], axis = 1)
match_gender = pd.concat([season_dataframe_men, season_dataframe_women], axis = 0).reset_index().drop("index", axis = 1)
average_game_gender =  pd.merge(average_games_perset_, match_gender, how = 'inner', )
average_game_gender.groupby("gender")["avg_games_per_set"].mean()
```


###
Answer
:

The average number of games per set in men's matches is **9.03**, and the average of this factor in women's match is **8.74**.
